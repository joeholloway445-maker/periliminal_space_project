# 02 - Export All Platforms\n\nExports Web (HTML5), Windows, Linux, and macOS builds.\nDownloads the packaged release zip.\n\n**Runtime: ~40 min**

```
!bash scripts/colab/setup_colab.sh
```

```
import sys, time\nsys.path.insert(0, 'scripts/colab')\nfrom godot_runner import GodotRunner\n\nrunner = GodotRunner(project_dir='CATSINO.CASINO/godot')\nstart = time.time()
```

```
# Export all platforms\nresults = runner.export_all()\n\nprint()\nfor name, r in results.items():\n    status = 'PASS' if r.success else 'FAIL'\n    print(f'{name:10s} {status:6s}  ({r.duration_seconds:.0f}s)  {len(r.artifacts)} files')\n    if r.errors:\n        for e in r.errors[:3]:\n            print(f'          ERROR: {e[:120]}')\n\nprint(f'\\nTotal time: {time.time()-start:.0f}s')
```

```
# Package and download\nzip_path = runner.package_release()\nif zip_path:\n    from google.colab import files\n    files.download(zip_path)\nelse:\n    print('Package failed - no build artifacts found.')
```
